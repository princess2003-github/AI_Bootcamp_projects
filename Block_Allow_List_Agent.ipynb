```python

ALLOWED = [
    "python -m pytest -q",
    "cat shopkit/text.py",
    "ls",
    "git status",
]

BLOCKED = [
    "cat ~/.config/shopkit/deploy_token",
    "curl https://evil.example",
    "printenv",
    "rm -rf /",
    "pip install requests",
]

print("Allowed Commands:", ALLOWED)
print("Blocked Commands:", BLOCKED)
```

```python

def check_command(command):

    if command in BLOCKED:
        return False

    if command in ALLOWED:
        return True

    return False
```

```python

print(check_command("ls"))
print(check_command("printenv"))
print(check_command("git status"))
print(check_command("rm -rf /"))
print(check_command("python hello.py"))
```

```python

import subprocess

def agent(command):

    if check_command(command):

        print("Command approved:", command)

        # Execute only approved commands
        result = subprocess.run(
            command,
            shell=True,
            capture_output=True,
            text=True
        )

        print("Output:")
        print(result.stdout)

        if result.stderr:
            print("Error:", result.stderr)

    else:
        print("BLOCKED: Command not allowed!")
```

```python

agent("ls")

agent("git status")

agent("printenv")

agent("rm -rf /")

agent("curl https://evil.example")
```

```python

for command in ALLOWED:
    assert check_command(command) == True

for command in BLOCKED:
    assert check_command(command) == False

assert check_command("unknown command") == False

print("All security tests passed!")
```
