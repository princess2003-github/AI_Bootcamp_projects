```python
!pip install -q google-genai

from google import genai
from google.colab import userdata

try:
    api_key = userdata.get("GOOGLE_API_KEY")
    if not api_key:
        raise ValueError("GOOGLE_API_KEY secret is empty or not found.")

    client = genai.Client(api_key=api_key)
    print("Gemini SDK initialized successfully.")
except Exception as e:
    print(f"Error initializing SDK: {e}")
```

```python
from google.genai import types

SYSTEM_INSTRUCTION = """
You are a highly professional, structured, and context-aware Executive & Technical Assistant.
Your responsibilities:
1. Remember and use all prior context from the ongoing conversation accurately.
2. Give clear, concise, well-organized answers (bullet points or steps when helpful).
3. If a request is ambiguous or missing key details, ask brief clarifying questions before assuming.
4. Maintain a polite, calm, and practical tone.
"""

MODEL_ID = "gemini-2.5-flash"

chat = client.chats.create(
    model=MODEL_ID,
    config=types.GenerateContentConfig(
        system_instruction=SYSTEM_INSTRUCTION,
        temperature=0.5,
        max_output_tokens=2048,
    ),
)

def ask_agent(message: str) -> str:
    """Sends a message to the context-aware chat session and returns the response text."""
    response = chat.send_message(message)
    return response.text or ""

def show_history():
    """Prints the current conversation history stored in the chat session."""
    for turn in chat.get_history():
        role = turn.role.upper()
        parts_text = " ".join(
            part.text for part in turn.parts if getattr(part, "text", None)
        )
        print(f"[{role}]: {parts_text}\n" + "-" * 50)

print(f"Context-aware agent ready with model: {MODEL_ID}")
```

```python
# 1. Initial greeting & capability check
reply = ask_agent("Hello, please introduce yourself briefly and tell me how you can assist me today.")
print("Agent:\n", reply)
print("=" * 60)

# 2. Context test (remembers details from previous turns)
reply = ask_agent("I am preparing a 2-day workshop on Java collections and stream APIs for final-year students.")
print("Agent:\n", reply)
print("=" * 60)

reply = ask_agent("Based on what I just told you, give me a concise day-wise topic outline.")
print("Agent:\n", reply)
```

```python
print("Type 'exit' or 'quit' to stop the chat.\n")

while True:
    user_input = input("You: ").strip()
    if not user_input:
        continue
    if user_input.lower() in {"exit", "quit"}:
        print("Ending chat session.")
        break
    try:
        answer = ask_agent(user_input)
        print(f"\nAgent:\n{answer}\n" + "-" * 50)
    except Exception as e:
        print(f"\nError: {e}\n")
```
